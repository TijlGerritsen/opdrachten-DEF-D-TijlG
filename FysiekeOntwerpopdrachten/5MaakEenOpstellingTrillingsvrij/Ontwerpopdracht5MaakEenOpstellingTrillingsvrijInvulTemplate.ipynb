## Opdracht 2b: Opsplitsen taak 2: afleiding ontwerpeis

```python
# Bereken hier je benodigde kantelfrequentie
def kantelfrequentie(f):
    f0 = np.sqrt(f**2/4)
    return f0
print(kantelfrequentie(5))
```

```python
# Voeg hier je plot in.

def amplitudeoverdracht(f0, f):
    omega0 = 2*np.pi*f0
    omega = 2*np.pi*f
    verhouding = omega**2 / omega0**2
    noemer = 1 - verhouding
    return np.abs(1/noemer)
f = np.linspace(0,20,100)
omega = f*np.pi*2
plt.figure()
plt.plot(omega,amplitudeoverdracht(kantelfrequentie(5), f))
plt.xscale('log')
plt.yscale('log')
plt.xlabel('Hoekfrequentie log(rad/s)')
plt.ylabel('Amplitudeoverdracht log(|Am/Av|)')
plt.savefig('amplitudeoverdracht.png')
```

```python
# Bereken hier de minimale Delta-x waarbij je aan de eis van de opdracht voldoet.
g = 9.81
def deltax(f0):
    omega0 = 2*np.pi*f0
    delta_x =  g / omega0**2
    return delta_x

minimale_deltax = deltax(kantelfrequentie(5))
print(minimale_deltax)
```

## Opdracht 3: Plot de grafiek.

```python
# Voeg hier je plot in. Lees goed de opdracht wat er in de grafiek moet staan!
# Het is ook handig om je grafiek op te slaan, doe dat door met plt.savefig().
def C(m, delta_x):
    C = m*g/delta_x
    return C
m = np.linspace(0,10,100)
plt.figure()
plt.plot(m, C(m, minimale_deltax), label='1x Minimale Delta-x')
plt.plot(m, C(m, 2*minimale_deltax), label='2x Minimale Delta-x')
plt.plot(m, C(m, 5*minimale_deltax), label='5x Minimale Delta-x')
plt.xlabel('Massa (kg)')
plt.ylabel('Veerconstante (N/m)')
plt.legend()
print(C(0.5, 2*minimale_deltax))
print(C(0.400,minimale_deltax))
plt.savefig('veerconstante.png')
```

```python
# Voeg hier je plot in.
def amplitudeoverdracht(f0, f):
    omega0 = 2*np.pi*f0
    omega = 2*np.pi*f
    verhouding = omega**2 / omega0**2
    noemer = 1 - verhouding
    return np.abs(1/noemer)
def eigenfrequentie(delta_x):
    omega0 = np.sqrt(g/delta_x)
    frequentie = omega0/(2*np.pi)
    return frequentie

f = np.linspace(0.1,10,1000)
omega = f*np.pi*2


x = [1.3*2*np.pi, 3.8*2*np.pi, 5.1*2*np.pi]
y = [2.55/1.51, 0.86/4.86, 0.35/2.64]

xerr = 0.1
yerr = np.zeros(len(y))

for i in range(len(y)):
    yerr[i] = y[i]/10


plt.figure()
plt.plot(omega,amplitudeoverdracht(kantelfrequentie(5), f))
plt.plot(omega,amplitudeoverdracht(eigenfrequentie(0.09),f))
plt.plot(1.3*2*np.pi,2.55/1.51, 'o', label='1 hz')
plt.plot(3.8*2*np.pi, 0.86/4.86, 'o', label='3 hz')
plt.plot(5.1*2*np.pi,0.35/2.64, 'o', label='5 hz')
plt.errorbar(x,y,xerr=xerr,yerr=yerr, fmt='o', label='Foutmarge', ecolor='red', capsize=5)
plt.xscale('log')
plt.yscale('log')
plt.xlabel('Hoekfrequentie log(rad/s)')
plt.ylabel('Amplitudeoverdracht log(|Am/Av|)')
plt.legend(['Kantelfrequentie', 'Eigenfrequentie bij Delta-x = 0.09m', '1 Hz', '3 Hz', '5 Hz'])
plt.savefig('amplitudeoverdracht_met_foutmarge.png')
```
